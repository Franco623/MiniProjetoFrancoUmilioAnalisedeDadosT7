​Carregar​ ​a ​​base​​Varejo.csv ​​com​ ​pandas​ ​e ​​mostrar:​​número​ ​de​ ​registros,​​colunas

```python
 #importando biblioteca pandas
import pandas as pd
import numpy as np
# Carregando base de dados.csv
# Com sep=';' a saida mostrava - Quantidade de colunas: 1
# Usando sep=',' para otimizar a importação do csv e mostrar o correto - Quantidade de colunas: 14

#df_varejo = pd.read_csv("Base_Varejo.csv", sep = ":")
df_varejo = pd.read_csv("/content/Base_Varejo.csv", sep = ",")
#  Mostra a quantidade de registros e colunas
print("\n--- DIMENSÕES DA BASE ---")
print(f"Quantidade de registros: {df_varejo.shape[0]}")
print(f"Quantidade de colunas: {df_varejo.shape[1]}")

# Mostra os nomes das colunas
print("\n--- Nomes das colunas ---")
print(df_varejo.columns.tolist())

# Mostra os tipos de dados
print("\n--- TIPOS DE DADOS ---")
print(df_varejo.dtypes)

# Mostra as primeiras linhas
#print ("\n--- PRIMEIRAS LINHAS DA BASE ---")
#print(df_varejo.head())
```

Segunda parte do trabalho:  
​Verificar​ ​e​ ​reportar​ ​ao​ ​menos​ ​dois​ ​problemas​ ​básicos:​ ​valores​ ​nulos​ ​por​ ​coluna,​
​duplicatas e possíveis inconsistências (ex.: datas inválidas ou categorias vazias).​

```python
valores_nulos = df_varejo.isnull().sum()
print(valores_nulos)
df_varejo.rename(columns={
    'Unnamed: 10': 'Sem_Categoria_1',
    'Unnamed: 11': 'Sem_Categoria_2',
    'Unnamed: 12': 'Sem_Categoria_3',
    'Unnamed: 13': 'Sem_Categoria_4'
}, inplace=False)
```

A saida acima detectou 4 colunas Unnamed (sem nome)com altos valores nulos. Olhando o arquivo  Base_Varejo.csv, percebi que as linhas Unnamed (sem nome) são textos continuação da coluna anterior. Segue um código para mostrar o conteudo das colunas Unnamed . Como uso LibreOffice em Debian 12, pode ser que baixando o arquivo csv da origine, uma malformação de configuração aconteceu.


```python

# Exibe as colunas sem nome
print("\n--- COLUNAS SEM NOME ---")

colunas_sem_nome = [
    coluna for coluna in df_varejo.columns
    if coluna.startswith("Sem_Categoria:")
]

print("\nQuantidade de valores preenchidos por coluna:")
print(df_varejo[colunas_sem_nome].notna().sum())

print("\nExemplos dos valores encontrados:")
print(df_varejo[colunas_sem_nome].head())
```

```python

# Reconstruir os nomes completos dos produtos
# e deletar as colunas com valores nulos

print("\n--- CORREÇÃO DOS NOMES DOS PRODUTOS ---")

colunas_produto = [
    "PR_NOME",
    "Sem_Categoria_1",
    "Sem_Categoria_2",
    "Sem_Categoria_3",
    "Sem_Categoria_4"
]

# Juntar as partes dos nomes, ignorando campos vazios
df_varejo["PR_NOME"] = (
    df_varejo[colunas_produto]
    .fillna("")
    .astype(str)
    .agg(" ".join, axis=1)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

# Excluir as colunas auxiliares após reconstruir os nomes
df_varejo = df_varejo.drop(
    columns=colunas_produto[1:]
)

print("Quantidade de colunas após a correção:",
      df_varejo.shape[1])

print("\nExemplos dos nomes reconstruídos:")
print(df_varejo["PR_NOME"].head(15).to_string(index=False))
```

Foram analisadas todas as colunas categóricas da base. Não foram encontradas categorias vazias nas variáveis CL_GENERO, CL_SEG, PR_CAT e PR_NOME. Registro Duplicados

```python
print("\n2. REGISTROS DUPLICADOS")
duplicados = df_varejo.duplicated().sum()
print(f"Quantidade de registros duplicados: {duplicados}")
```

```python
print("\n2. Removendo duplicatas...")

qtd_duplicados = df_varejo.duplicated().sum()

df_varejo = df_varejo.drop_duplicates()

print(f"Duplicatas removidas: {qtd_duplicados}")
```

```python
df_varejo['DATA'] = pd.to_datetime(
    df_varejo['DATA'],
    format='%d/%m/%y',
    errors='coerce'
)

datas_invalidas = df_varejo['DATA'].isna().sum()

print(f"Datas inválidas: {datas_invalidas}")

print(df_varejo['DATA'].head())
```
