```python
 #importando biblioteca pandas
import pandas as pd
import numpy as np
# Carregando base de dados.csv
# Com sep=';' a saida mostrava - Quantidade de colunas: 1
# Usando sep=',' para otimizar a importação do csv e mostrar o correto - Quantidade de colunas: 14

#df_varejo = pd.read_csv("Base_Varejo.csv", sep = ":")
df_varejo = pd.read_csv("/content/Base_Varejo.csv", sep = ",")
#  Mostra a quantidade de registros e colunas
print("\n--- DIMENSÕES DA BASE ---")
print(f"Quantidade de registros: {df_varejo.shape[0]}")
print(f"Quantidade de colunas: {df_varejo.shape[1]}")

# Mostra os nomes das colunas
print("\n--- Nomes das colunas ---")
print(df_varejo.columns.tolist())

# Mostra os tipos de dados
print("\n--- TIPOS DE DADOS ---")
print(df_varejo.dtypes)

# Mostra as primeiras linhas
#print ("\n--- PRIMEIRAS LINHAS DA BASE ---")
#print(df_varejo.head())

```

Segunda parte do trabalho:  
​Verificar​ ​e​ ​reportar​ ​ao​ ​menos​ ​dois​ ​problemas​ ​básicos:​ ​valores​ ​nulos​ ​por​ ​coluna,​
​duplicatas e possíveis inconsistências (ex.: datas inválidas ou categorias vazias).​

Mundando os nomes Unnamed para Sem_Categoria usando un for para percorrer as colunas é un if else para achar "Unnamed" e tracar para "Sem_Categoria" o código a seguir exibe antes e depois


```python
print("--- Antes ---")
print(df_varejo.columns.tolist())

novas_colunas = []
for coluna in df_varejo.columns:
    if "Unnamed" in str(coluna):
        novas_colunas.append("Sem_Categoria")
    else:
        novas_colunas.append(coluna)

df_varejo.columns = novas_colunas

print("\n--- Depois ---")
print(df_varejo.columns.tolist())

```

 Para melhorar a interpretação dos dados e atender aos requisitos de tratamento de categorias "Unnamed"(sem nome), as colunas foram renomeadas "Sem_Categoria"


Segue um print da tabela com nomes corrigidos e soma os valores nulos existentes.


```python
# exibe a base de dados corrigida
print(df_varejo)

# soma valores nulos e exibe

valores_nulos = df_varejo.isnull().sum()
print(valores_nulos)
```

Foram analisadas todas as colunas . Não foram encontradas categorias vazias nas variáveis CL_GENERO, CL_SEG, PR_CAT e PR_NOME. Entretanto, foram identificadas colunas Sem_Categoria, contendo grande quantidade de valores "null" ou em algumas linhas de trecho de texto das linhas anteriores. Estando com valores nulos podem ser deletadas não antes de reconstruir as colunas PR_NOME para uma melhor leitura da base de dados



```python
# Reconstruir os nomes completos dos produtos
print("\n--- CORREÇÃO DOS NOMES DOS PRODUTOS ---")

colunas_produto = [
    "PR_NOME",
    "Sem_Categoria",
    #"Unnamed: 11",
    #"Unnamed: 12",
    #"Unnamed: 13"
]

# Juntar as partes dos nomes, ignorando campos vazios
df_varejo["PR_NOME"] = (
    df_varejo[colunas_produto]
    .fillna("")
    .astype(str)
    .agg(" ".join, axis=1)
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

# Excluir as colunas auxiliares após reconstruir os nomes
df_varejo = df_varejo.drop(
    columns=colunas_produto[1:]
)

print("Quantidade de colunas após a correção:",
      df_varejo.shape[1])

#print("\nExemplos dos nomes reconstruídos:")
#print(df_varejo["PR_NOME"].head(15).to_string(index=False))
```

Reconstruidas as colunas  os valores null são deletados a seguir
as colunas passam de 14 para 10  sem valores nulos dectettados.
